```python
from qdrant_client import QdrantClient
from qdrant_client.http.models import Filter

def delete_points_by_document_ids(client, collection_name, document_ids):
    """
    Delete points from a Qdrant collection based on a list of document IDs.

    Args:
        client (QdrantClient): The initialized Qdrant client.
        collection_name (str): The name of the collection.
        document_ids (list): A list of document IDs to delete.

    Returns:
        dict: The response from the Qdrant server.
    """
    # Create a filter to match the document_ids
    filter_criteria = Filter(
        must=[
            {"key": "document_id", "match": {"any": document_ids}}
        ]
    )

    # Delete points matching the filter
    response = client.delete(
        collection_name=collection_name,
        points_selector=filter_criteria
    )
    return response

# Usage Example
client = QdrantClient(
    url="http://localhost:6333",  # Replace with your Qdrant instance URL
    api_key="QdrantVAsfhF8nGPtyleJKVkt2TBI2bqQ4bSjgnajNtOLwE2Y9YWxnZFrItRBE53"        # Replace with your actual API key
)

collection_name = "faq_data"      # Replace with your collection name
document_ids = ["faq_th_28", "faq_th_29"]  # List of document IDs to delete

# Call the function to delete the specified points
response = delete_points_by_document_ids(client, collection_name, document_ids)

# Print the response
print(response)
```
